Data Cleaning

```python
df.fillna(0,inplace = True)
df.isnull().sum()
```

```python
df['Goals Scored'].dtype
```

```python
df['Goals Scored'] = df['Goals Scored'].replace('-',0)
df['Goals Scored'] = df['Goals Scored'].replace("N,A",0)
df['Assists Provided'] = df['Assists Provided'].replace('-',0)
df['Assists Provided'] = df['Assists Provided'].replace("N,A",0)
df[' Appearances'] = df[' Appearances'].replace('-',0)
df[' Appearances'] = df[' Appearances'].replace("N,A",0)
df['Dribbles per 90'] = df['Dribbles per 90'].replace('-',0)
df['Dribbles per 90'] = df['Dribbles per 90'].replace("N,A",0)
df['Interceptions per 90'] = df['Interceptions per 90'].replace('-',0)
df['Interceptions per 90'] = df['Interceptions per 90'].replace("N,A",0)
df['Total Duels Won per 90'] = df['Total Duels Won per 90'].replace('-',0)
df['Total Duels Won per 90'] = df['Total Duels Won per 90'].replace("N,A",0)
df['Save Percentage'] = df['Save Percentage'].replace('-',0)
df['Save Percentage'] = df['Save Percentage'].replace("N,A",0)
df['Clean Sheets'] = df['Clean Sheets'].replace('-',0)
df['Clean Sheets'] = df['Clean Sheets'].replace("N,A",0)
df['Tackles per 90'] = df['Tackles per 90'].replace("N,A",0)
df['Tackles per 90'] = df['Tackles per 90'].replace("-",0)
```

```python
df.columns
```

```python
df['Tackles per 90'].unique()
```

```python
df['Goals Scored'] = pd.to_numeric(df['Goals Scored'],errors = 'coerce')
df[' Appearances'] = pd.to_numeric(df[' Appearances'],errors = 'coerce')
df['Assists Provided'] = pd.to_numeric(df['Assists Provided'],errors = 'coerce')
df['Dribbles per 90'] = pd.to_numeric(df['Dribbles per 90'],errors = 'coerce')
df['Interceptions per 90'] = pd.to_numeric(df['Interceptions per 90'],errors = 'coerce')
df['Total Duels Won per 90'] = pd.to_numeric(df['Total Duels Won per 90'],errors = 'coerce')
df['Save Percentage'] = pd.to_numeric(df['Save Percentage'],errors = 'coerce')
df['Clean Sheets'] = pd.to_numeric(df['Clean Sheets'],errors = 'coerce')
df['Tackles per 90'] = pd.to_numeric(df['Tackles per 90'],errors = 'coerce')
df
```

Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dx = df.groupby('Nationality')[['Goals Scored','Assists Provided']].sum()
dx = dx.sort_values('Goals Scored',ascending = False)
dx = dx.head()
dx
```

```python
sns.barplot(x = dx['Assists Provided'],y = 'Nationality',data = dx,hue = 'Nationality',palette = 'muted',width = 0.6,edgecolor = 'black')
plt.title('Assists of the countries with most goals in the World Cup',fontsize = 15)
plt.show()
print('Analysis 1:-')
print('   - 1.France along with England shows the most number of assists provided in the world cup.')
print('   - 2.while other 3 countries show the assists provided above 7.')
```

```python
d2 = df.sort_values('Clean Sheets',ascending = False)
d2 = d2.head()
d2
```

```python
sns.barplot(x = 'Save Percentage',y = 'Player Name',data = d2,hue = 'Player Name',palette = 'magma',width = 0.6,edgecolor = 'black')
plt.title('Most Saves')
plt.show()
print('Analysis 2:-')
print('   - 1.From the graph it comes to know that all top 5 goalkeepers with most clean sheets have save percentage above 0.5')
print('   - 2.The gragh shows that except Davis Epassy,from Matt Turner to Munir Mohamedi the graph increases by above 10%')
```

```python
df.drop_duplicates(inplace = True)
```

```python
brand = df.groupby('National Team Kit Sponsor').count()
brand = brand.sort_values('Nationality',ascending = False)
brand = brand.head(6)
brand
```

```python
sns.barplot(x = 'National Team Kit Sponsor',y = 'Nationality',data = brand,hue = 'National Team Kit Sponsor',palette = 'viridis',edgecolor = 'black',width = 0.6)
plt.ylabel('Player Count')
plt.xlabel('National Team Kit Sponsor')
plt.title('Player Distribution By National Team Kit Sponsor')
plt.show()
print('Analysis 3:-')
print('   - 1.The bar plot shows that the NIKE has sponsored the most players.')
print('   - 2.The graph also interprets that KAPPA,HUMMELS and HUMMEL all has sponsored players number below 50.')
print('   - 3.The Brands Adidas and PUMA has sponsored the players in the 150 - 200 range.')
```

```python
sns.scatterplot(x = 'Dribbles per 90',y = 'Total Duels Won per 90',data = df[df['Dribbles per 90'] < 11],color = 'orange',edgecolor = 'black')
plt.title('Dribbles per 90 vs Duels Won per 90',fontsize = 15)
plt.show()
print('Analysis 4:-')
print('   - 1.From the graph it seems that Most of the players have Both Dribbles in range 0 - 2 and Duels in range 0 - 10.')
print('   - 2.Also graph interprets that only few players have Dribbles above 2 and Duels above 10.')
print('   - 3.It seems from the graph that there is one outlier which has Dribbles per 90 above 10.')
```

```python
d3 = df.groupby('Club').count()
d3 = d3.sort_values('Nationality',ascending = False)
d3 = d3.head(10)
d3
```

```python
sns.barplot(x = 'Nationality',y = 'Club',data = d3,hue = 'Club',palette = 'Set2',edgecolor = 'black')
plt.title('Clubs With Most Players In the Tournament',fontsize = 15)
plt.xlabel('Players Count',fontsize = 12)
plt.ylabel('Clubs',fontsize = 12)
plt.show()
print('Analysis 5:-')
print('   - 1.From the barplot it seems Clubs Bayern Munich,Manchester City and FC Barcelona has most, 16 - players through out the teams in the world cup.')
print('   - 2.Also From the graph it comes to know that Clubs like AL Hilal,PSG,Real Madrid,chelsea and Athletico Madrid all has total of 12 players playing in the Tournament.')
print('   - 3.While the clubs Manchester United and Al Sadd had player count in the range 12 - 14.')
```

```python
sns.heatmap(df.corr(numeric_only = True),annot = True,fmt = '.0f',cmap = 'Blues')
plt.title('Correlation between different numeric attributes',fontsize = 15)
plt.show()
print('Analysis 5:-')
print('   - 1.From the Heat map it seems that Save percentage and Clean Sheets have the biggest correlation in number - 1.')
print('   - 2.The Assists Provided,Goals Scored and Appearances how correlation above 0.2.')
```

```python
sns.boxplot(x = 'Dribbles per 90',data = df[df['Dribbles per 90'] < 4],width = 0.8,color = 'Skyblue')
plt.title('Dribbles Per 90',fontsize = 15)
plt.show()
print('Analysis 6:-')
print('   - 1.From the graph is comes to know that most of the players have the Tackles per 90 in range 0 - 1.5 .')
print('   - 2.While above range 3 there are some of the outliers.')
```

```python
sns.boxplot(x = 'Tackles per 90',data = df[df['Tackles per 90'] < 6])
plt.title('Tackles per 90',fontsize = 15)
plt.show()
print('Analysis 7:-')
print('   - 1.From the graph is comes to know that most of the players have the Tackles per 90 in range 0 - 2.')
print('   - 2.While above range 5 there are some of the outliers.')
```

```python
ds = df.groupby('FIFA Ranking')[['Goals Scored','Assists Provided']].sum()
ds = ds.sort_values('FIFA Ranking',ascending = False)
ds
```

```python
sns.lineplot(x = 'FIFA Ranking',y = ds['Goals Scored'],data = ds,marker = 'o',color = 'red')
plt.title('Goals Scored Vs Fifa Ranking',fontsize = 15)
plt.show()
print('Analysis 8:-')
print('   - 1.From the Lineplot is comes to know that most of the teams in the Ranking range 1-10 has goals scored above 7.')
print('   - 2.Also from the graph it seems there is decrease in the number of goals scored by the team as their ranking decreases.')
```
